# Pre-Processing and Cleaning of Data

```python
import polars as pl
from pathlib import Path
```

# Load Raw Data from Public CSVs

```python
dir_path = Path().cwd().parent.parent
if (dir_path / Path('data/raw_data')).exists():
    raw_data_path = dir_path / Path('data/raw_data')
else:
    (dir_path / Path('data/raw_data')).mkdir(parents=True, exist_ok=True)
    raw_data_path = dir_path / Path('data/raw_data')
```

### Cleaning the Location related Columns

```python
# Sub-setting the crash DataFrame for only the columns I need
location_df = crash_data[['COLLISION ID', 'CRASH DATE', 'CRASH TIME', 'LATITUDE',
       'LONGITUDE', 'LOCATION', 'NUMBER OF PERSONS INJURED',
       'NUMBER OF PERSONS KILLED', 'CONTRIBUTING FACTOR VEHICLE 1',
       'CONTRIBUTING FACTOR VEHICLE 2']]
```

```python
display(location_df.describe())
```

```python
# Check for missing values 
print("Missing values in Location Data:")
print(location_df.null_count())
```

```python
# Drop all the rows where location is missing
location_df = location_df.drop_nulls(subset=['LOCATION'])

location_df = location_df.filter(
    (pl.col("LATITUDE").is_between(40.5774, 45.01585))  &
        (pl.col("LONGITUDE").is_between(-74.2591, -73.7004))
)
location_df = location_df.drop("LOCATION")
```

```python
# Cleaning of Contributing Factors

rename_dict ={
    'Cell Phone (hand-held)': 'Cell Phone (Hand-Held)',
    'Cell Phone (hand-Held)': 'Cell Phone (Hand-Held)',
    'Cell Phone (hands-free)': 'Cell Phone (Hands-Free)',
    'Drugs (illegal)': 'Drugs (Illegal)', 
    'Illnes': 'Illness'
}

location_df = location_df.drop_nulls(
    subset=["CONTRIBUTING FACTOR VEHICLE 1", "CONTRIBUTING FACTOR VEHICLE 2"]
    ).filter(
        (
            (pl.col("CONTRIBUTING FACTOR VEHICLE 1") != "Unspecified") |
            (pl.col("CONTRIBUTING FACTOR VEHICLE 2") != "Unspecified")
        )
        &
        (pl.col("CONTRIBUTING FACTOR VEHICLE 1") != "80")
        &
        (pl.col("CONTRIBUTING FACTOR VEHICLE 2") != "80")
        &
        (pl.col("CONTRIBUTING FACTOR VEHICLE 1") != "1")
        &
        (pl.col("CONTRIBUTING FACTOR VEHICLE 2") != "1")
    ).with_columns(
        pl.col("CONTRIBUTING FACTOR VEHICLE 1").replace(rename_dict),
        pl.col("CONTRIBUTING FACTOR VEHICLE 2").replace(rename_dict),
    )
```

```python
# fill nulls with 'UNKNOWN' or 0 

location_df = location_df.with_columns(
    pl.col("NUMBER OF PERSONS INJURED").fill_null(0.0),
    pl.col("NUMBER OF PERSONS KILLED").fill_null(0.0)
)

print("Location Data after cleaning:")
print(location_df.null_count())
```

```python
location_df.head()
```

```python
import datetime as dt
# Handle Types

location_df = location_df.with_columns(
    pl.col("CRASH DATE").str.strptime(pl.Date, format="%m/%d/%Y", strict=False),
    
    # Convert CRASH_TIME to time (as HH:MM)
    pl.col("CRASH TIME").str.strptime(pl.Time, format="%H:%M", strict=False),
    
    # Convert numeric columns to integer
    pl.col("NUMBER OF PERSONS INJURED").cast(pl.Int32),
    pl.col("NUMBER OF PERSONS KILLED").cast(pl.Int32),
)

location_df.head()
```

```python
# Creating Encoding Data
location_df = location_df.with_columns(
    pl.when(pl.col("NUMBER OF PERSONS KILLED") > 0)
      .then(pl.lit("Fatal"))
      .when(pl.col("NUMBER OF PERSONS INJURED") >= 3)
      .then(pl.lit("Major Injury"))
      .when(pl.col("NUMBER OF PERSONS INJURED") > 0)
      .then(pl.lit("Minor Injury"))
      .otherwise(pl.lit("No Injury"))
      .alias("ACCIDENT SEVERITY")
)

print(location_df.head())
```

```python
location_df.head()
```

## Attempt to tie Locations (LAT and LONG) to Boroughs

```python
# Try to find the Borough for each location entry
import geopandas as gpd

points_df = location_df.to_pandas()

points_gdf = gpd.GeoDataFrame(
    points_df,
    geometry=gpd.points_from_xy(
        points_df["LONGITUDE"],
        points_df["LATITUDE"]
    ),
    crs="EPSG:4326"
)

# Load the shapefile of boroughs (assuming it contains a 'BOROUGH' column)
shapefile_path = dir_path.parent  / Path('Borough Boundaries/geo_borough.shp')
boroughs_gdf = gpd.read_file(shapefile_path)

if boroughs_gdf.crs is None:
    boroughs_gdf.set_crs(epsg=2263, inplace=True)

# Ensure the shapefile uses the same CRS as the points
boroughs_gdf = boroughs_gdf.to_crs(epsg=4326)
# Perform a spatial join to find which borough each point is in
result_gdf = gpd.sjoin(points_gdf, boroughs_gdf, how='inner', predicate='within')

result_gdf.head()
```

```python
location_df = result_gdf.drop(columns=['boro_code', 'shape_area', 'shape_leng', 'geometry', 'index_right'])
location_df.rename(columns={'boro_name': 'BOROUGH'}, inplace=True)
location_df.head()
```

# Save clean data with boroughs in csvs

```python
location_df.to_csv(str(dir_path / Path('data/clean_data/location_data.csv')), index=False)
```

# Subsetting Data to focus on relevant columns


```python

person_detail_df = person_data.select(['COLLISION ID', 'PERSON AGE', 'PERSON TYPE', 'PERSON SEX', 'BODILY INJURY', 'EJECTION'])

person_detail_df.head()
```

```python
avg_age = (
    person_detail_df
    .select(pl.col("PERSON AGE").mean())
    .item()
)
avg_age = round(avg_age)

person_detail_df = person_detail_df.with_columns(
    pl.col("PERSON AGE").fill_null(avg_age)
)

person_detail_df.head()
```

```python
print("Before the conversion:")
print(person_detail_df.schema)

person_detail_df = person_detail_df.with_columns(
    pl.col("PERSON AGE").cast(pl.Int32)
)

print("After the conversion:")
print(person_detail_df.schema)

person_detail_df.head()
```

```python
# Change the Person Sex Column
person_detail_df = person_detail_df.with_columns(
    pl.col("PERSON SEX")
      .replace({
          "M": "Male",
          "F": "Female",
          "U": "Unknown",
      })
      .fill_null("UNKNOWN")
)

print(
    person_detail_df
    .select(pl.col("PERSON SEX").value_counts())
)
```

```python
# Drop the columns where the bodily injury is not described
person_detail_df = person_detail_df.filter(
    pl.col("BODILY INJURY") != "Unknown"
)

person_detail_df.select(
    pl.col("BODILY INJURY").value_counts()
)
```

```python
person_detail_df.columns
```

Create a super dataframe with Location and Additional Person data

```python
# Replace spaces with underscores in all column names
location_df = pl.from_pandas(location_df)
merge_df = location_df.join(
    person_detail_df,
    on="COLLISION ID",
    how="inner"
)

print("Location DF shape:", location_df.shape)
print("Person DF shape:", person_detail_df.shape)
print("Merged DF shape:", merge_df.shape)

merge_df.head(10)
```

```python

# Fill nulls for specific columns
merge_df = merge_df.with_columns([
    pl.col("EJECTION").fill_null("Unknown"),
    pl.col("BODILY INJURY").fill_null("Unknown"),
    pl.col("PERSON AGE").fill_null(0),
    pl.col("PERSON TYPE").fill_null("Unknown"),
    pl.col("PERSON SEX").fill_null("Unknown"),
])

# Replace 'Does Not Apply' with 'Unknown' in string columns
string_cols = ["EJECTION", "BODILY INJURY", "PERSON TYPE", "PERSON SEX"]
merge_df = merge_df.with_columns([
    pl.when(pl.col(c) == "Does Not Apply")
      .then(pl.lit("Unknown"))
      .otherwise(pl.col(c))
      .alias(c)
    for c in string_cols
])
```
